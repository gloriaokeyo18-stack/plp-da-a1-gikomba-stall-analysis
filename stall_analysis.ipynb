#Introdution

This analysis examines the sales performance and inventory position of Wanjiku's fresh-produce stall at Gikomba Market over a 14-day period. Using core Python, the analysis evaluated daily sales trends, week-on-week growth, sales performance across different days, inventory requirements and product gross margins.The objective is to provde Wanjiku with practial, data-driven recommendations to improve business performance and inventory management.

```python

daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

##Task 1: Overall Sales Performance

In this section I calculate the total, average, highest and lowest daily sales over the 14-day period. I also identify the corresponding day names and week numbers using list indexing.

```python

# Calculate overall sales metrics

total_sales = sum(daily_sales)

average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)

lowest_sales = min(daily_sales)


# Identify the positions of the highest and lowest sales

highest_index = daily_sales.index(highest_sales)

lowest_index = daily_sales.index(lowest_sales)


# Identify the corresponding days

best_day = days[highest_index]

worst_day = days[lowest_index]


# Identify the corresponding weeks

best_week = (highest_index // 7) + 1

worst_week = (lowest_index // 7) + 1


# Display the results

print("OVERALL SALES PERFORMANCE")
print("-------------------------")
print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Daily Sales: KES {highest_sales:,.2f}")
print(f"Best Day: {best_day}, Week {best_week}")
print(f"Lowest Daily Sales: KES {lowest_sales:,.2f}")
print(f"Worst Day: {worst_day}, Week {worst_week}")
```


## Task 2: Week-on-Week Sales Growth

In this section, I define a reusable function called week_totals() that uses list slicing to calculate the total sales for each week. I then calculate the percentage change in sales between Week 1 and Week 2 to establish whether the business experienced growth or a decline.


```python

def week_totals(sales):
    """
    Calculate total sales for Week 1 and Week 2.

    Args:
        sales: A list containing 14 daily sales figures.

    Returns:
        A tuple containing Week 1 and Week 2 sales totals.
    """
    week1_total = sum(sales[:7])
    week2_total = sum(sales[7:])

    return week1_total, week2_total


# Call the function and unpack its results

week1, week2 = week_totals(daily_sales)


# Calculate week-on-week growth percentage

growth_percentage = ((week2 - week1) / week1) * 100


# Display the results

print("WEEK-ON-WEEK SALES PERFORMANCE")
print("--------------------------------")
print(f"Week 1 Sales: KES {week1:,.2f}")
print(f"Week 2 Sales: KES {week2:,.2f}")
print(f"Sales Growth: {growth_percentage:.2f}%")
```


## Task 3: Daily Sales Classification

In this section, I define a function called classify_day() to classify daily sales as Strong, Normal or Weak based on their deviation from the average daily sales. I then apply the function to all 14 days and use a dictionary to count the number of days in each category.


```python

def classify_day(amount, average):
    """
    Classify daily sales based on their difference from average sales.

    Args:
        amount: The sales amount for a particular day.
        average: The average daily sales.

    Returns:
        Strong, Normal or Weak depending on the sales performance.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


# Create a dictionary to count each category

sales_count = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}


# Classify each day's sales

daily_classifications = []

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_classifications.append(category)
    sales_count[category] += 1


# Display daily classifications

print("DAILY SALES CLASSIFICATION")
print("--------------------------")

for day, amount, category in zip(days, daily_sales, daily_classifications):
    print(f"{day}: KES {amount:,.2f} - {category}")


# Display category counts

print("\nSALES CATEGORY SUMMARY")
print("----------------------")

for category, count in sales_count.items():
    print(f"{category}: {count} days")
```


## Task 4: Weekday versus Weekend Sales Performance

In this section, I separate daily sales into weekdays (Monday to Friday) and weekends (Saturday and Sunday) using list slicing. I calculate the average sales for each group and determine the percentage difference to identify which period generates higher average sales.


```python

# Separate weekday and weekend sales

weekdays = daily_sales[:5] + daily_sales[7:12]

weekends = daily_sales[5:7] + daily_sales[12:14]


# Calculate average sales for each group

weekday_average = sum(weekdays) / len(weekdays)

weekend_average = sum(weekends) / len(weekends)


# Calculate percentage difference

percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100


# Display the results

print("WEEKDAY VS WEEKEND PERFORMANCE")
print("--------------------------------")
print(f"Average Weekday Sales: KES {weekday_average:,.2f}")
print(f"Average Weekend Sales: KES {weekend_average:,.2f}")
print(f"Percentage Difference: {percentage_difference:.2f}%")

if weekend_average > weekday_average:
    print("Finding: Weekend sales are higher than weekday sales.")
elif weekend_average < weekday_average:
    print("Finding: Weekday sales are higher than weekend sales.")
else:
    print("Finding: Weekday and weekend sales are equal.")
```


## Task 5: Inventory Analysis and Gross Margins

In this section, I use the stock dictionary to identify products whose available quantities are at or below their reorder levels. I define a reusable function to return these products and calculate the gross margin percentage for each product using its cost and selling price.


```python

def products_to_reorder(stock):
    """
    Identify products whose stock quantity is at or below
    their specified reorder level.

    Args:
        stock: A dictionary containing product inventory details.

    Returns:
        A list of product names that need restocking.
    """
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


# Identify products requiring restocking

reorder_products = products_to_reorder(stock)


# Calculate gross margins for all products

gross_margins = {}

for product, details in stock.items():
    cost = details["cost"]
    price = details["price"]

    margin = ((price - cost) / price) * 100

    gross_margins[product] = margin


# Display products requiring restocking

print("INVENTORY REORDER ALERT")
print("-----------------------")

if reorder_products:
    for product in reorder_products:
        print(
            f"{product}: {stock[product]['qty']} kg in stock "
            f"(Reorder level: {stock[product]['reorder']} kg)"
        )
else:
    print("No products currently require restocking.")


# Display gross margins

print("\nGROSS MARGIN ANALYSIS")
print("---------------------")

for product, margin in gross_margins.items():
    print(f"{product}: {margin:.2f}%")
```


## Task 6: Stall Performance Report

```python

# STALL PERFORMANCE REPORT

print("=" * 55)
print("          WANJIKU'S STALL PERFORMANCE REPORT")
print("=" * 55)

# SECTION 1: OVERALL SALES PERFORMANCE

print("\n1. OVERALL SALES PERFORMANCE")
print("-" * 55)

print(f"Total Sales:             KES {total_sales:,.2f}")
print(f"Average Daily Sales:     KES {average_sales:,.2f}")
print(f"Highest Daily Sales:    KES {highest_sales:,.2f}")
print(f"Best Sales Day:          {best_day}, Week {best_week}")
print(f"Lowest Daily Sales:     KES {lowest_sales:,.2f}")
print(f"Worst Sales Day:         {worst_day}, Week {worst_week}")


# SECTION 2: WEEK-ON-WEEK PERFORMANCE

print("\n2. WEEK-ON-WEEK PERFORMANCE")
print("-" * 55)

print(f"Week 1 Sales:            KES {week1:,.2f}")
print(f"Week 2 Sales:            KES {week2:,.2f}")
print(f"Sales Growth:            {growth_percentage:.2f}%")


# SECTION 3: DAILY SALES CLASSIFICATION

print("\n3. DAILY SALES CLASSIFICATION")
print("-" * 55)

for category, count in sales_count.items():
    print(f"{category:<15}: {count} days")


# SECTION 4: WEEKDAY VS WEEKEND PERFORMANCE

print("\n4. WEEKDAY VS WEEKEND PERFORMANCE")
print("-" * 55)

print(f"Average Weekday Sales:   KES {weekday_average:,.2f}")
print(f"Average Weekend Sales:   KES {weekend_average:,.2f}")
print(f"Weekend Sales Advantage: {percentage_difference:.2f}%")


# SECTION 5: INVENTORY REORDER ALERT

print("\n5. INVENTORY REORDER ALERT")
print("-" * 55)

if reorder_products:
    for product in reorder_products:
        details = stock[product]
        print(
            f"{product:<15}: {details['qty']} kg available "
            f"(Reorder at {details['reorder']} kg)"
        )
else:
    print("No products require restocking.")


# SECTION 6: GROSS MARGIN ANALYSIS

print("\n6. GROSS MARGIN ANALYSIS")
print("-" * 55)

for product, margin in gross_margins.items():
    print(f"{product:<15}: {margin:.2f}%")

print("\n" + "=" * 55)
print("                 END OF REPORT")
print("=" * 55)
```
